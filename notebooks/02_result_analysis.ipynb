# Notebook 02: Simulation Result Analysis

Analyzes the simulation output across the progressive autonomy modes:
- Baseline (Random)
- Heuristic Agent
- Supervised Agent
- Adaptive Agent

Generates comparisons with the published results from Rodrigo Furlan de Assis et al. (2026).

```python
import os
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# Load final metrics table
table_path = Path('../outputs/tables/table_final_metrics.csv')
if table_path.exists():
    df_summary = pd.read_csv(table_path)
    display(df_summary)
else:
    print('table_final_metrics.csv not found. Run simulation first.')
```

```python
# Plot trajectory comparison
metrics_dir = Path('../outputs/metrics')
if metrics_dir.exists():
    plt.figure(figsize=(10, 6))
    for f in metrics_dir.glob('metrics_*.csv'):
        mode_name = f.stem.replace('metrics_', '').capitalize()
        m_df = pd.read_csv(f)
        plt.plot(m_df['event_id'], m_df['average_mi'], label=mode_name, linewidth=2)
    plt.xlabel('Event Number')
    plt.ylabel('Average Mixing Index')
    plt.title('Mixing Index Progression by Mode')
    plt.legend()
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.show()
```
